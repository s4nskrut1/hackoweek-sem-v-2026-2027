## 1. Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Vectors
A vector is an ordered collection of numbers. In machine learning, a vector can represent the features of one data sample.

```python
v1 = np.array([2, 3, 4])
v2 = np.array([1, 5, 2])

print('Vector 1:', v1)
print('Vector 2:', v2)
print('Vector addition:', v1 + v2)
print('Vector magnitude:', np.linalg.norm(v1))
```

### ML Intuition
A feature vector such as `[age, income, experience]` can represent one observation. Machine-learning algorithms perform mathematical operations on these vectors to make predictions.

## 3. Matrices
A matrix is a rectangular arrangement of numbers. A dataset can naturally be represented as a matrix where rows are observations and columns are features.

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

print('Matrix A:')
print(A)
print('Shape:', A.shape)
print('Transpose:')
print(A.T)
```

## 4. Dot ProductThe dot product multiplies corresponding elements of two vectors and adds the results. It is fundamental to linear models and neural networks.

```python
v1 = np.array([2, 3, 4])
v2 = np.array([1, 5, 2])

dot_product = np.dot(v1, v2)
manual = (2*1) + (3*5) + (4*2)

print('Dot product:', dot_product)
print('Manual calculation:', manual)
```

### ML ConnectionA linear model commonly computes a weighted sum such as `w · x + b`. The dot product is the `w · x` part of this calculation.

## 5. Eigenvalues and EigenvectorsFor a matrix, an eigenvector is a direction that remains the same direction after the matrix transformation, while its eigenvalue describes the amount of scaling in that direction. This is an intuition-level view used in methods such as PCA.

```python
M = np.array([[4, 1],
              [2, 3]])

eigenvalues, eigenvectors = np.linalg.eig(M)

print('Matrix:')
print(M)
print('\nEigenvalues:')
print(eigenvalues)
print('\nEigenvectors:')
print(eigenvectors)
```

### ML Connection
Eigenvalues and eigenvectors help identify important directions in data. Principal Component Analysis (PCA), for example, uses eigenvector-based ideas to find directions that capture high variance.

# Calculus

## 6. DerivativesA derivative measures how quickly a function changes with respect to its input. In machine learning, derivatives help determine how a model's output or loss changes when a parameter changes.

```python
def f(x):
    return x**2

def derivative(x):
    return 2*x

x = np.linspace(-5, 5, 200)
plt.figure(figsize=(7, 5))
plt.plot(x, f(x), label='f(x) = x²')
plt.plot(x, derivative(x), label="f'(x) = 2x")
plt.xlabel('x')
plt.ylabel('Value')
plt.title('Function and Its Derivative')
plt.legend()
plt.grid(True)
plt.show()
```

For `f(x) = x²`, the derivative is `f'(x) = 2x`. The derivative tells us the slope of the function at a particular point.

## 7. GradientsA gradient extends the idea of a derivative to functions with multiple variables. It is a vector containing partial derivatives and points in the direction of greatest increase.

```python
def gradient(x, y):
    # f(x,y) = x^2 + y^2
    return np.array([2*x, 2*y])

point = (2, 3)
print('Point:', point)
print('Gradient at point:', gradient(*point))
```

### ML ConnectionFor a model with many parameters, the gradient tells us how the loss changes with respect to each parameter. Optimization algorithms such as gradient descent use this information to update model parameters.

## 8. Gradient Descent IntuitionGradient descent moves parameters in the opposite direction of the gradient to reduce the loss.

```python
x = 4.0
learning_rate = 0.1
history = [x]

for _ in range(20):
    grad = 2*x  # derivative of x^2
    x = x - learning_rate * grad
    history.append(x)

print('Starting x: 4.0')
print('Final x:', round(x, 6))

plt.figure(figsize=(7, 4))
plt.plot(history, marker='o')
plt.xlabel('Iteration')
plt.ylabel('x value')
plt.title('Gradient Descent Moving Toward the Minimum')
plt.grid(True)
plt.show()
```

## 9. Chain RuleThe chain rule is used to differentiate a composition of functions. If `y = f(g(x))`, then `dy/dx = (dy/dg)(dg/dx)`. It is the mathematical foundation of backpropagation.

```python
# Example: y = (3x + 2)^2
x = 2

# Let u = 3x + 2 and y = u^2
u = 3*x + 2
dy_du = 2*u
du_dx = 3
dy_dx = dy_du * du_dx

print('x =', x)
print('u = 3x + 2 =', u)
print('dy/du =', dy_du)
print('du/dx =', du_dx)
print('dy/dx using chain rule =', dy_dx)
```

## 10. Backpropagation IntuitionNeural networks contain multiple layers of functions. During training, the loss is propagated backward through these layers. The chain rule allows the network to calculate how much each weight contributed to the final error. These gradients are then used to update the weights.

```python
# A tiny computational example
x = 2.0
w = 3.0
b = 1.0
target = 10.0

# Forward pass: y = wx + b
y = w*x + b
loss = (y - target)**2

# Backward pass using the chain rule
dL_dy = 2*(y - target)
dy_dw = x
dL_dw = dL_dy * dy_dw

print('Prediction:', y)
print('Loss:', loss)
print('Gradient of loss with respect to weight:', dL_dw)
```
